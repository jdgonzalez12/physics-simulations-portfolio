```python
import numpy as np
import matplotlib.pyplot as plt


# Problem setup

a = 110 # Diffusion coefficient (mm^2/s)
L = 50 # Rod length in millimeters
tiempo = 10 # Total simulation time in seconds
nodos = 20 # Number of nodes/points on the discretized rod

# Initialization

dx = L / (nodos-1) # Spacing between nodes
dt = 0.5 * dx**2 / a  # Time step computed with the CFL stability criterion, to ensure the finite-difference method doesn't diverge
t_nodos = int(tiempo/dt) + 1 # Number of time nodes

u = np.zeros(nodos) + 20 # Rod initially at 20 degrees

# Boundary conditions

u[0] = 100 # Temperature at the first node
u[-1] = 100 # Temperature at the last node


# Live plot setup

fig, axis = plt.subplots()

pcm = axis.pcolormesh([u], cmap=plt.cm.jet, vmin=0, vmax=100)
plt.colorbar(pcm, ax=axis)
axis.set_ylim([-2, 3])

# Simulation

counter = 0

while counter < tiempo :

    w = u.copy() # Copy the u array to avoid modifying it while it's being used

    for i in range(1, nodos - 1): # loop over interior nodes

        u[i] = dt * a * (w[i - 1] - 2 * w[i] + w[i + 1]) / dx ** 2 + w[i] # approximate the second spatial derivative

    counter += dt

    print("t: {:.3f} [s], Average temperature= {:.2f} Celsius".format(counter, np.average(u)))

    # Update the live plot

    pcm.set_array([u])
    axis.set_title("Distribution at t = {:.3f} [s].".format(counter))
    plt.pause(0.01) # Pause to update the plot in real time


plt.show()
```
