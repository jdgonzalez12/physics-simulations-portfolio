CALL OPTION EXAMPLE (NVIDIA-style)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load the index-based grid
df = pd.read_csv("data/grid_V_ij_call.csv", header=None)
V = df.values

# Load the grid with real (S, t) axes
df_real = pd.read_csv("data/grid_V_S_t_call.csv")
t_values = df_real.iloc[1:, 0].astype(float).values
S_values = df_real.columns[1:].astype(float).values
V_real = df_real.iloc[1:, 1:].astype(float).values

# Plot type (a): evolution of V[i][j]
plt.figure(figsize=(8, 5))
for j in range(0, V.shape[0], V.shape[0] // 20):
    plt.plot(V[j], label=f"{j}")
plt.title("Evolution of V[i][j] with respect to changes in the underlying asset price (Call)")
plt.xlabel("Spatial index i")
plt.ylabel("V[i][j]")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("data/mesh_evolution_call.png")
plt.show()

# Plot type (b): surface V(S,t)
t_grid, S_grid = np.meshgrid(t_values, S_values, indexing='ij')  # t on X, S on Y
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(t_grid, S_grid, V_real, cmap='viridis')
ax.set_title("Surface V(S, t) -- Call")
ax.set_xlabel("Time (days)")              # X axis = time
ax.set_ylabel("Underlying asset price S")  # Y axis = space
ax.set_zlabel("Option value V(S,t)")       # Z axis = value
plt.savefig("data/surface_V_call.png")
plt.show()

# Plot type (c): heatmap
plt.figure(figsize=(10, 6))
plt.imshow(V_real, aspect='auto', origin='lower', extent=[S_values.min(), S_values.max(), t_values.min(), t_values.max()], cmap='plasma')
plt.colorbar(label='V(S,t)')
plt.title("Heatmap V(S,t) -- Call")
plt.xlabel("Underlying asset price S, in [0,Smax]")
plt.ylabel("Time (days)")
plt.tight_layout()
plt.savefig("data/heatmap_V_call.png")
plt.show()
```
