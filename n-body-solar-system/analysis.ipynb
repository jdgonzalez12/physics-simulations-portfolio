# Plots

Visualizations of the N-body gravity simulations (see the `.cpp` files in this
folder). All trajectories use the corrected Velocity Verlet integrator
(position update with the old acceleration, then velocity update with the
average of the old and newly recomputed acceleration).

```python
import matplotlib.pyplot as plt
import pandas as pd

def read_data(filename):
    data = pd.read_csv(filename)
    return data['x1'], data['y1'], data['x2'], data['y2']

files = [
    ("data/two_body_mass_ratio_high.csv", "m1 > m2"),
    ("data/two_body_mass_ratio_equal.csv", "m1 = m2"),
    ("data/two_body_mass_ratio_low.csv", "m1 < m2")
]

plt.figure(figsize=(18, 5))

for i, (filename, title) in enumerate(files):
    x1, y1, x2, y2 = read_data(filename)

    plt.subplot(1, 3, i + 1)
    plt.plot(x1, y1, label="Body 1", color='blue')
    plt.plot(x2, y2, label="Body 2", color='orange')

    # mark the starting point
    plt.plot(x1.iloc[0], y1.iloc[0], 'bo', label="Body 1 start")
    plt.plot(x2.iloc[0], y2.iloc[0], 'ro', label="Body 2 start")

    plt.title(title)
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.grid()
    plt.legend()

plt.tight_layout()
plt.savefig("data/two_body_orbits.png", dpi=150)
plt.show()
```

## Part 1.4 - Sun-Earth exact mass ratio

Two-body case using the exact Sun/Earth mass ratio (332946).

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv("data/sun_earth.csv")

# Build the plot
plt.figure(figsize=(8, 6))
plt.plot(df["x1"], df["y1"], label="Sun", color='orange')
plt.plot(df["x2"], df["y2"], label="Earth", color='blue')

# mark starting positions
plt.scatter(df["x1"].iloc[0], df["y1"].iloc[0], color='red', s=50, label="Sun start")
plt.scatter(df["x2"].iloc[0], df["y2"].iloc[0], color='green', s=50, label="Earth start")

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Sun-Earth trajectories")
plt.legend()
plt.grid()
plt.axis("equal")
plt.savefig("data/sun_earth.png", dpi=150)
plt.show()
```

## Part 2 - Sun-Earth-Moon three-body system

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/sun_earth_moon.csv")

plt.figure(figsize=(10, 8))

# Trajectories
plt.plot(df["x1"], df["y1"], label="Sun", color="orange", linewidth=2)
plt.plot(df["x2"], df["y2"], label="Earth", color="blue", linewidth=1.5)
plt.plot(df["x3"], df["y3"], label="Moon", color="black", linewidth=1)

# Starting points
plt.scatter(df["x1"].iloc[0], df["y1"].iloc[0], color="orange", label="Sun start", s=40)
plt.scatter(df["x2"].iloc[0], df["y2"].iloc[0], color="blue", label="Earth start", s=40)
plt.scatter(df["x3"].iloc[0], df["y3"].iloc[0], color="black", label="Moon start", s=40)

# Aspect ratio tweak to make the ellipticity visible
plt.gca().set_aspect(1.5)  # try 1.2, 2, etc.
# plt.xlim(-2e11, 2e11)
# plt.ylim(-1.2e11, 1.2e11)

plt.title("Sun-Earth-Moon trajectories")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("data/sun_earth_moon.png", dpi=150)
plt.show()
```

## Part 3 - Full solar system (N-body)

Sun plus all 8 planets, simulated for the full ~165-year Neptune orbital
period so every planet completes at least one full revolution.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/solar_system.csv")

plt.figure(figsize=(12, 10))

# Planet trajectories
plt.plot(df["x1"], df["y1"], label="Sun", color="orange")
plt.plot(df["x2"], df["y2"], label="Mercury", color="gray")
plt.plot(df["x3"], df["y3"], label="Venus", color="goldenrod")
plt.plot(df["x4"], df["y4"], label="Earth", color="blue")
plt.plot(df["x5"], df["y5"], label="Mars", color="red")
plt.plot(df["x6"], df["y6"], label="Jupiter", color="saddlebrown")
plt.plot(df["x7"], df["y7"], label="Saturn", color="gold")
plt.plot(df["x8"], df["y8"], label="Uranus", color="lightseagreen")
plt.plot(df["x9"], df["y9"], label="Neptune", color="blueviolet")

# Starting points
plt.scatter(df["x1"].iloc[0], df["y1"].iloc[0], color="orange", s=40, marker="o")
plt.scatter(df["x2"].iloc[0], df["y2"].iloc[0], color="gray", s=40, marker="o")
plt.scatter(df["x3"].iloc[0], df["y3"].iloc[0], color="goldenrod", s=40, marker="o")
plt.scatter(df["x4"].iloc[0], df["y4"].iloc[0], color="blue", s=40, marker="o")
plt.scatter(df["x5"].iloc[0], df["y5"].iloc[0], color="red", s=40, marker="o")
plt.scatter(df["x6"].iloc[0], df["y6"].iloc[0], color="saddlebrown", s=40, marker="o")
plt.scatter(df["x7"].iloc[0], df["y7"].iloc[0], color="gold", s=40, marker="o")
plt.scatter(df["x8"].iloc[0], df["y8"].iloc[0], color="lightseagreen", s=40, marker="o")
plt.scatter(df["x9"].iloc[0], df["y9"].iloc[0], color="blueviolet", s=40, marker="o")

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Solar system planet trajectories (with starting points)")
plt.legend()
plt.axis("equal")
plt.grid(True)
plt.tight_layout()
plt.savefig("data/solar_system.png", dpi=150)
plt.show()
```
